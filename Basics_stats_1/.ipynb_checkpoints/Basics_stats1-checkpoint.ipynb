### Data Visualization

```python
# histograms
numerical_col = df.select_dtypes(include = ['int64','float64']).columns
for col in numerical_col:
    plt.figure(figsize = (8,8))
    plt.hist(df[col].dropna(),bins = 10)
    plt.title(f"HISTOGRAM OF {col}")
    plt.xlabel(col)
    plt.ylabel("frequency")
    plt.show()
```

    1.Histograms shows how data values are distributed.
    2.symmetric distribution are close to normal.
    3. skewed distribution may requried transformation.

```python
# BOX PLOTS

for col1 in numerical_col:
    plt.figure(figsize = (6,6))
    plt.boxplot(df[col1].dropna())
    plt.title(f"BOX PLOT OF {col1}")
    plt.ylabel(col1)
    plt.show()
```

    boxplot is used for finding the outliers.
    outliers are present beyond the whiskers.
    some variables show large spread and variability.

```python
for col in numerical_col:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\nColumn: {col}")
    print("IQR:", IQR)
    print("Number of Outliers:", len(outliers))
```

```python
categorical_col = df.select_dtypes(include = ['object']).columns
print(categorical_col)
```

```python
for col in categorical_col:
    plt.figure(figsize = (6,6))
    df[col].value_counts().plot(kind='bar')
    plt.title(f'Frequency of {col}')
    plt.xlabel('Count')
    plt.ylabel(col)
    plt.show()
```

    1.Bar charts display the frequency of each category.
    2.Categories with higher bars occur more frequently.
    3.Uneven distributions may indicate class imbalance.

Key Insights:

    1. Numerical variables were analyzed using descriptive statistics.
    2. Histograms revealed the distribution patterns.
    3. Box plots identified potential outliers.
    4. Categorical variables were analyzed using frequency charts.
    5. These findings can support further predictive analysis.
