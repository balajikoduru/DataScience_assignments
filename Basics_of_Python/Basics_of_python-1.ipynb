### Exercise 1: Prime Numbers

write a Python program that checks whether a given number is prime or not. 
A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself

```python

n = int(input("Enter a number: "))

if n <= 1:
    print("Not a prime number")
else:
    is_prime = True

    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime number")
    else:
        print("Not a prime number")
        
```

### Exercise 2: Product of Random Numbers

Develop a Python program that generates two random numbers and asks the user to enter the product of these numbers. 
    The program should then check if the user's answer is correct and display an appropriate message.

```python
import random

# Generate two random numbers
num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

# Ask the user for the product
answer = int(input(f"What is {num1} × {num2}? "))

# Check the answer
if answer == num1 * num2:
    print("Correct! ")
else:
    print(f"Wrong! The correct answer is {num1 * num2}.")
```

### Exercise 3: Squares of Even/Odd Numbers

Create a Python script that prints the squares of all even or odd numbers within the range of 100 to 200. Choose either even or odd numbers and document your choice in the code.

```python
# Choice: Even numbers
# This program prints the squares of all even numbers from 100 to 200.

for num in range(100, 201):
    if num % 2 == 0:
        print(f"{num}² = {num ** 2}")
```

### Exercise 4:
✅ Problem Statement (Question): Hospital Billing System

A hospital charges its patients based on the following:
    
    Room charges per day: ₹2,000
    Doctor consultation fee (flat): ₹1,500
    Lab test charges: Based on the number of tests taken. Each test costs ₹300
    Medicine charges: Total cost of medicines provided
    Discount: If the total bill (before discount) exceeds ₹10,000, a 10% discount is applied
Write a Python program to calculate the final bill for a patient who:
    
    Stayed for 4 days
    Had 3 lab tests
    Had medicine charges of ₹2,400

```python
# Hospital Billing System

# Given values
days = 4
room_charge_per_day = 2000
doctor_fee = 1500
lab_tests = 3
lab_charge_per_test = 300
medicine_charges = 2400

# Calculate charges
room_charges = days * room_charge_per_day
lab_charges = lab_tests * lab_charge_per_test

# Calculate total bill before discount
total_bill = room_charges + doctor_fee + lab_charges + medicine_charges

# Apply discount
if total_bill > 10000:
    discount = total_bill * 0.10
else:
    discount = 0

# Calculate final bill
final_bill = total_bill - discount

# Display the bill
print("----- Hospital Bill -----")
print("Room Charges: ₹", room_charges)
print("Doctor Consultation Fee: ₹", doctor_fee)
print("Lab Test Charges: ₹", lab_charges)
print("Medicine Charges: ₹", medicine_charges)
print("Total Bill: ₹", total_bill)
print("Discount: ₹", discount)
print("Final Bill: ₹", final_bill)
```
