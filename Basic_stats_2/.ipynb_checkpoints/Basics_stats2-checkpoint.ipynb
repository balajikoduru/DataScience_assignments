2.Select only the columns relevant for billing: ['PatientID', 'Department', 'Doctor', 'BillAmount'].

```python
df[['PatientID','Department','Doctor','BillAmount']]
```

3.Drop administrative columns like ['ReceptionistID', 'CheckInTime'].

```python
df.drop(['ReceptionistID','CheckInTime'],axis = 1)
```

4.Use groupby to find total bill amount per department.

```python
df.groupby('Department')['BillAmount'].sum()
```

5.Remove duplicate patient records based on PatientID.

```python
df['PatientID'].duplicated().sum()
```

6.Fill missing BillAmount values with the mean bill amount.

```python
clean_data = df.drop_duplicates(subset = 'PatientID')
clean_data
```

```python
amount_mean = df['BillAmount'].mean()
amount_mean
```

```python
df['BillAmount'] = df['BillAmount'].fillna(amount_mean)
```

```python
clean_data
```

```python
billing_df = pd.read_csv('Billing_Data.csv')
```

7.Merge the billing dataset with patient dataset on PatientID.

```python
merged_df = pd.merge(clean_data, billing_df, on='PatientID')
```

```python
merged_df
```

    8.Concatenate an additional DataFrame that contains new patients for the current week (row-wise).
    9.Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise).

```python
new_patient_df = pd.DataFrame({
    'PatientID':[106,107],
    'Name':['Devi Vara Prasad','Jyothinadh'],
    'Department':['Neurology','Orthopedics'],
    'Doctor':['Dr. Sharma','Dr. Rao'],
    'BillAmount':[5000,6000],
    'ReceptionistID':[4,5],
    'CheckInTime':['2026-06-12','2026-06-13'],
    'InsuranceCovered':[2000,3000],
    'FinalAmount':[3000,3000]
})

updated_df = pd.concat([merged_df,new_patient_df],axis = 0,ignore_index = True)
updated_df
```
