```python

# 1.Create a list of 5 integers

nums = [20,30,40,50,60]
print("Before updating the list: ",nums)
#append() method is used for adding element at end of list

nums.append(70)
print("After append a 70 element in the list: ",nums)

#extend() method is used for adding multiple elements at end of list

nums.extend([80,90])
print("After extends the elements(80,90) in the list: ",nums)

#insert() method is used for insert element at specific place or specific index value in a list

nums.insert(0,10)
print("After insert a 10 value in the list: ",nums)

#remove() method is used for remove the specific element in the list

nums.remove(90)
print("After remove a 90 element in the list: ",nums)

#pop() method is used for delete the element at end of the list

nums.pop()
print("After pop a element in the list:",nums)

#index() method is used for to find the index value at specific element position

index_value = nums.index(50)
print("50 element index value is: ",index_value)

#count() method is used for count the how many elements in the list

count_values = nums.count(60)
print("How many elements in the list(60): ",count_values)

#sort() method is used for arrange the data in order 

nums.sort()
print("Sort the ascending order in list: ",nums)

#reverse() method is used for reverse the list

nums.reverse()
print("Revere the list: ",nums)

#clear() method is used for delete all elements in the list

nums.clear()
print("Clear a list: ",nums)
```

```python
# 6.checking for grade with marks

marks = int(input("Enter a your marks:"))

if marks >= 90:
    print("Your Grade is: A")
elif marks >= 75 and marks <= 89:
    print("Your Grade is: B")
elif marks >= 50 and marks <= 74:
    print("Your Grade is: C")
else:
    print("Fail")
```

```python
# 7. Check a number if it positive and even.if not positive its a zero or negative

number = int(input("Enter a number:"))

if number > 0:
    print("number is positive")
    if number % 2 == 0:
        print("its even number")
    else:
        print("its odd number")
elif number < 0:
    print("number is a negative")
else:
    print("it is a zero")
```

### Numpy

```python
# NUMPY
# 8. Create numpy array

# importing the numpy 

import numpy as np

# A scalar using np.array(5)

arr = np.array(5)
print(arr)
print(type(arr))
print(arr.ndim)
print(arr.shape)
print("******************************")
# A 1D array with values 1 to 5

arr1 = np.array([1,2,3,4,5])
print(arr1)
print(type(arr1))
print(arr1.ndim)
print(arr1.shape)
print("******************************")
# A 2D array (2x3) with values from 10 to 60 in steps of 10 

arr2 = np.array([[10,20,30],[40,50,60]])
print(arr2)
print(type(arr2))
print(arr2.ndim)
print(arr2.shape)
```

```python
# 10. Create a 2D numpy array of shape(3x3).Convert it into pandas data frame and add columns names

#import numpy and pandas

import numpy as np
import pandas as pd

# Create a 3x3 numpy array

arr = np.array([[10,20,30],[40,50,60],[70,80,90]])
print(arr)

print("****************************************")

# Convert numpy into pandas dataframe

df = pd.DataFrame(arr)
print(df)

print("****************************************")

# add Columns names is A,B,C

df.columns = ['A','B','C']
print(df)
```

###  Exploring Pandas

```python
# 11. Create a small Dataframe manually with 10 rows and columns

# import pandas

import pandas as pd

# Create a dataframe

df = pd.DataFrame({'Name':['koduru balaji','vijay','sandeep kishan','Sri Ram','Varun','Kumar','Bhuvan','Sai','Pavan','Ram'],
                   'Age':[22,23,22,18,26,25,34,21,27,24],
                   'City':['Nellore','Vijayawada','Kadapa','Krishna','Hyderabad','Guntur','Vizayanagaram','sanathnagar','Palanadu','chittor'],
                   'Salary':[25000,26000,28000,30000,32000,34000,36000,38000,40000,42000]
                  })
print(df)
print("#######################################")

# use info() and describe()

print(df.info())
print("#######################################")
print(df.describe())
print("#######################################")

#select 'Name' and 'City' columns

print(df[['Name','City']])
print("#######################################")

#Drop 'City' column

df = df.drop('City',axis = 1)
print(df)



#fill any missing values in 'Salary' column with the mean

print("How many missing values is: ",df['Salary'].isnull().sum())



#Remove any duplicated rows

print("How many duplicated rows: ",df.duplicated().sum())
```
