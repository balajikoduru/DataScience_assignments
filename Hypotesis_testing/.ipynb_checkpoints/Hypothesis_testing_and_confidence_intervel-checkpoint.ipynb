```python
import numpy as np
from scipy import stats
```

### Hypothesis testing

```python
# Hypothesis Testing for Bombay Hospitality Ltd.

import math
from scipy.stats import norm

# Given Data
sample_mean = 3050          # x̄
X_mean = 600                # Mean units produced
sigma_X = 25                # Standard deviation of units produced
n = 25                      # Sample size
alpha = 0.05               # Significance level

# Step 1: Calculate Theoretical Mean Cost
mu = 1000 + 5 * X_mean
print("Theoretical Mean Cost (μ):", mu)

# Step 2: Calculate Population Standard Deviation of Cost
sigma = 5 * sigma_X
print("Population Standard Deviation (σ):", sigma)

# Step 3: Calculate Test Statistic (Z)
z_stat = (sample_mean - mu) / (sigma / math.sqrt(n))
print("Z-Test Statistic:", round(z_stat, 4))

# Step 4: Critical Value for Right-Tailed Test
z_critical = norm.ppf(1 - alpha)
print("Critical Value:", round(z_critical, 4))

# Step 5: P-value
p_value = 1 - norm.cdf(z_stat)
print("P-value:", round(p_value, 6))

# Step 6: Decision
if z_stat > z_critical:
    print("\nDecision: Reject the Null Hypothesis")
else:
    print("\nDecision: Fail to Reject the Null Hypothesis")

# Step 7: Conclusion
if p_value < alpha:
    print("Conclusion: There is sufficient evidence to support the claim that weekly operating costs are higher than the model suggests.")
else:
    print("Conclusion: There is not enough evidence to support the claim that weekly operating costs are higher than the model suggests.")
```

### Confidence Intervel

```python
# sample data

data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]

# Sample Size 

n = len(data)

# Sample mean

mean = np.mean(data)

# Sample Standard Deviation

s = np.std(data)

print("sample size (n):",n)
print("Sample Mean:",round(mean,4))
print("Smaple Standard Deviation:",round(s,4))
```

```python
# 99 % Confidence Interval using Sample Standard Deviation

confidence = 0.99

alpha = 1 - confidence


# t critical value

t_critical = stats.t.ppf(1 - alpha/2,df=n-1)

# Margin of Error

margin_error_t = t_critical * (s / np.sqrt(n))

# confidence interval

lower_t = mean - margin_error_t
upper_t = mean + margin_error_t

print('Lower Limit:',round(lower_t,4))
print('Upper Limit:',round(upper_t,4))
```

```python
# 99% confidence Interval using known Papulation SD

# populationn Standard deviation

sigma = 0.2

# z critical value

z_critical = stats.norm.ppf(1 - alpha / 2)

# Margin of Error

margin_error_z = z_critical * (sigma / np.sqrt(n))

# confidence interval

lower_z = mean - margin_error_z
upper_z = mean + margin_error_z


print("Lower limit:",round(lower_z,4))
print("Upper Limit:",round(upper_z,4))
```
