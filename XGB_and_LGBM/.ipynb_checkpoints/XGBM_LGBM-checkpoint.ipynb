# Data Preprocessing:

```python
# Select all columns except Outcome as input features
X = df.drop('Outcome', axis=1)

# Select Outcome as the target variable
y = df['Outcome']


# Display the shape of the input features
print("Shape of X:", X.shape)

# Display the shape of the target
print("Shape of y:", y.shape)


# Display the feature names
print("\nInput features:")
print(X.columns.tolist())


# Display the target values
print("\nTarget values:")
print(y.unique())
```

```python
# Check the number of missing values in every column
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Import SimpleImputer from sklearn
from sklearn.impute import SimpleImputer


# Create the imputer
# Median values will be used if any missing values are present
imputer = SimpleImputer(strategy='median')


# Fit the imputer on X and transform the data
X_imputed = imputer.fit_transform(X)


# Convert the result back into a DataFrame
X_imputed = pd.DataFrame(
    X_imputed,
    columns=X.columns
)


# Check whether missing values remain
print("Missing values after imputation:")
print(X_imputed.isnull().sum())
```

```python
# Display the data type of every feature
print("Data types of input features:")
print(X_imputed.dtypes)
```

```python
# Create a list of features where zero values may represent
# missing or invalid measurements
zero_check_columns = [
    'Glucose',
    'BloodPressure',
    'SkinThickness',
    'Insulin',
    'BMI'
]


# Count zero values in each selected feature
zero_counts = (X_imputed[zero_check_columns] == 0).sum()

print("Number of zero values:")
print(zero_counts)
```

```python
# Create a copy of the input data
X_clean = X_imputed.copy()


# Replace zero values with NaN for selected medical measurements
X_clean[zero_check_columns] = X_clean[zero_check_columns].replace(
    0,
    np.nan
)


# Display the newly created missing values
print("Missing values after treating zero measurements as missing:")
print(X_clean.isnull().sum())
```

```python
# Create a new median imputer
medical_imputer = SimpleImputer(strategy='median')


# Apply median imputation to the selected medical features
X_clean[zero_check_columns] = medical_imputer.fit_transform(
    X_clean[zero_check_columns]
)


# Check the final missing value count
print("Missing values after median imputation:")
print(X_clean.isnull().sum())
```

```python
# Check the number of duplicate records
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

```python
# Remove duplicate records if they are present
df_clean = df.drop_duplicates().reset_index(drop=True)


# Display the new shape
print("Shape after removing duplicates:")
print(df_clean.shape)
```

```python
# Create the final feature matrix
X = X_imputed.copy()

# Create the target variable
y = df['Outcome'].copy()


# Display the final feature matrix
print("Final feature data:")
print(X.head())


# Display the target variable
print("\nFinal target data:")
print(y.head())


# Display the final dimensions
print("\nFinal X shape:", X.shape)
print("Final y shape:", y.shape)
```

# Building Predictive Models:

```python
# Split the dataset into training and testing data
# 80 percent of the data is used for training
# 20 percent of the data is used for testing
# random_state ensures that the same split is obtained every time
# stratify maintains the same proportion of diabetic and non diabetic cases
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the shape of the training data
print("Training feature shape:", X_train.shape)
print("Training target shape:", y_train.shape)

# Display the shape of the testing data
print("Testing feature shape:", X_test.shape)
print("Testing target shape:", y_test.shape)
```

```python
# Create the LightGBM classification model
lgbm_model = LGBMClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=4,
    num_leaves=31,
    random_state=42,
    verbosity=-1
)

# Train the LightGBM model using the training data
lgbm_model.fit(X_train, y_train)

# Generate predictions on the testing data
lgbm_pred = lgbm_model.predict(X_test)

# Display the first few predictions
print("LightGBM predictions:")
print(lgbm_pred[:10])
```

```python
# Calculate LightGBM accuracy
lgbm_accuracy = accuracy_score(y_test, lgbm_pred)

# Calculate LightGBM precision
lgbm_precision = precision_score(y_test, lgbm_pred)

# Calculate LightGBM recall
lgbm_recall = recall_score(y_test, lgbm_pred)

# Calculate LightGBM F1 score
lgbm_f1 = f1_score(y_test, lgbm_pred)

# Display the evaluation results
print("LightGBM Performance")
print("--------------------")
print("Accuracy :", lgbm_accuracy)
print("Precision:", lgbm_precision)
print("Recall   :", lgbm_recall)
print("F1 Score :", lgbm_f1)
```

```python
# Generate a detailed classification report
print("LightGBM Classification Report:")
print(classification_report(y_test, lgbm_pred))
```

```python
# Calculate the confusion matrix
lgbm_cm = confusion_matrix(y_test, lgbm_pred)

# Display the confusion matrix
print("LightGBM Confusion Matrix:")
print(lgbm_cm)

# Create a visual confusion matrix
ConfusionMatrixDisplay(
    confusion_matrix=lgbm_cm,
    display_labels=["Non diabetic", "Diabetic"]
).plot()

# Add a title to the plot
plt.title("LightGBM Confusion Matrix")

# Display the plot
plt.show()
```

```python
# Create the XGBoost classification model
xgb_model = XGBClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric='logloss'
)

# Train the XGBoost model using the training data
xgb_model.fit(X_train, y_train)

# Generate predictions on the testing data
xgb_pred = xgb_model.predict(X_test)

# Display the first few predictions
print("XGBoost predictions:")
print(xgb_pred[:10])
```

```python
# Calculate XGBoost accuracy
xgb_accuracy = accuracy_score(y_test, xgb_pred)

# Calculate XGBoost precision
xgb_precision = precision_score(y_test, xgb_pred)

# Calculate XGBoost recall
xgb_recall = recall_score(y_test, xgb_pred)

# Calculate XGBoost F1 score
xgb_f1 = f1_score(y_test, xgb_pred)

# Display the evaluation results
print("XGBoost Performance")
print("-------------------")
print("Accuracy :", xgb_accuracy)
print("Precision:", xgb_precision)
print("Recall   :", xgb_recall)
print("F1 Score :", xgb_f1)
```

```python
# Generate a detailed classification report
print("XGBoost Classification Report:")
print(classification_report(y_test, xgb_pred))
```

```python
# Calculate the confusion matrix
xgb_cm = confusion_matrix(y_test, xgb_pred)

# Display the confusion matrix values
print("XGBoost Confusion Matrix:")
print(xgb_cm)

# Create a visual confusion matrix
ConfusionMatrixDisplay(
    confusion_matrix=xgb_cm,
    display_labels=["Non diabetic", "Diabetic"]
).plot()

# Add a title
plt.title("XGBoost Confusion Matrix")

# Display the plot
plt.show()
```

```python
# Create a DataFrame containing the performance of both models
results = pd.DataFrame({
    'Model': ['LightGBM', 'XGBoost'],
    'Accuracy': [lgbm_accuracy, xgb_accuracy],
    'Precision': [lgbm_precision, xgb_precision],
    'Recall': [lgbm_recall, xgb_recall],
    'F1 Score': [lgbm_f1, xgb_f1]
})

# Display the comparison table
print(results)
```

```python
# Round the evaluation metrics to four decimal places
results = results.round(4)

# Display the final comparison
print(results)
```

```python
# Set the model column as the index
plot_data = results.set_index('Model')

# Create a bar chart for comparing all evaluation metrics
plot_data.plot(
    kind='bar',
    figsize=(10, 6)
)

# Add the chart title
plt.title("LightGBM vs XGBoost Performance")

# Add the x axis label
plt.xlabel("Model")

# Add the y axis label
plt.ylabel("Score")

# Set the y axis range
plt.ylim(0, 1)

# Rotate x axis labels
plt.xticks(rotation=0)

# Display the legend
plt.legend()

# Display the plot
plt.show()
```

```python
# Create Stratified K Fold cross validation
# The dataset is divided into 5 folds
# Each fold maintains the class distribution
skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
# Perform five fold cross validation for LightGBM
lgbm_cv_scores = cross_val_score(
    lgbm_model,
    X,
    y,
    cv=skf,
    scoring='accuracy'
)

# Display the accuracy obtained from each fold
print("LightGBM Cross Validation Scores:")
print(lgbm_cv_scores)

# Calculate the average cross validation accuracy
print("\nLightGBM Mean CV Accuracy:",
      lgbm_cv_scores.mean())

# Calculate the standard deviation
print("LightGBM CV Standard Deviation:",
      lgbm_cv_scores.std())
```

```python
# Perform five fold cross validation for XGBoost
xgb_cv_scores = cross_val_score(
    xgb_model,
    X,
    y,
    cv=skf,
    scoring='accuracy'
)

# Display the accuracy obtained from each fold
print("XGBoost Cross Validation Scores:")
print(xgb_cv_scores)

# Calculate the average cross validation accuracy
print("\nXGBoost Mean CV Accuracy:",
      xgb_cv_scores.mean())

# Calculate the standard deviation
print("XGBoost CV Standard Deviation:",
      xgb_cv_scores.std())
```

```python
# Define the parameters that will be tested for LightGBM
lgbm_param_grid = {
    'n_estimators': [100, 200],
    'learning_rate': [0.01, 0.05, 0.1],
    'max_depth': [3, 4, 5],
    'num_leaves': [15, 31]
}

# Create the GridSearchCV object
lgbm_grid = GridSearchCV(
    estimator=LGBMClassifier(
        random_state=42,
        verbosity=-1
    ),
    param_grid=lgbm_param_grid,
    cv=5,
    scoring='f1',
    n_jobs=-1
)

# Perform hyperparameter tuning using the training data
lgbm_grid.fit(X_train, y_train)

# Display the best parameters found
print("Best LightGBM Parameters:")
print(lgbm_grid.best_params_)

# Display the best cross validation F1 score
print("\nBest LightGBM CV F1 Score:")
print(lgbm_grid.best_score_)
```

```python
# Get the best LightGBM model
best_lgbm = lgbm_grid.best_estimator_

# Generate predictions using the tuned model
best_lgbm_pred = best_lgbm.predict(X_test)

# Calculate the evaluation metrics
best_lgbm_accuracy = accuracy_score(y_test, best_lgbm_pred)
best_lgbm_precision = precision_score(y_test, best_lgbm_pred)
best_lgbm_recall = recall_score(y_test, best_lgbm_pred)
best_lgbm_f1 = f1_score(y_test, best_lgbm_pred)

# Display the tuned LightGBM results
print("Tuned LightGBM Performance")
print("--------------------------")
print("Accuracy :", best_lgbm_accuracy)
print("Precision:", best_lgbm_precision)
print("Recall   :", best_lgbm_recall)
print("F1 Score :", best_lgbm_f1)
```

```python
# Define the parameters that will be tested for XGBoost
xgb_param_grid = {
    'n_estimators': [100, 200],
    'learning_rate': [0.01, 0.05, 0.1],
    'max_depth': [3, 4, 5],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

# Create the GridSearchCV object
xgb_grid = GridSearchCV(
    estimator=XGBClassifier(
        random_state=42,
        eval_metric='logloss'
    ),
    param_grid=xgb_param_grid,
    cv=5,
    scoring='f1',
    n_jobs=-1
)

# Perform hyperparameter tuning
xgb_grid.fit(X_train, y_train)

# Display the best parameters
print("Best XGBoost Parameters:")
print(xgb_grid.best_params_)

# Display the best cross validation F1 score
print("\nBest XGBoost CV F1 Score:")
print(xgb_grid.best_score_)
```

```python
# Get the best XGBoost model
best_xgb = xgb_grid.best_estimator_

# Generate predictions using the tuned model
best_xgb_pred = best_xgb.predict(X_test)

# Calculate the evaluation metrics
best_xgb_accuracy = accuracy_score(y_test, best_xgb_pred)
best_xgb_precision = precision_score(y_test, best_xgb_pred)
best_xgb_recall = recall_score(y_test, best_xgb_pred)
best_xgb_f1 = f1_score(y_test, best_xgb_pred)

# Display the tuned XGBoost results
print("Tuned XGBoost Performance")
print("-------------------------")
print("Accuracy :", best_xgb_accuracy)
print("Precision:", best_xgb_precision)
print("Recall   :", best_xgb_recall)
print("F1 Score :", best_xgb_f1)
```

```python
# Create a DataFrame containing the tuned model results
tuned_results = pd.DataFrame({
    'Model': ['Tuned LightGBM', 'Tuned XGBoost'],
    'Accuracy': [
        best_lgbm_accuracy,
        best_xgb_accuracy
    ],
    'Precision': [
        best_lgbm_precision,
        best_xgb_precision
    ],
    'Recall': [
        best_lgbm_recall,
        best_xgb_recall
    ],
    'F1 Score': [
        best_lgbm_f1,
        best_xgb_f1
    ]
})

# Round the values for better readability
tuned_results = tuned_results.round(4)

# Display the final comparison
print(tuned_results)
```

```python
# Set Model as the index
tuned_plot_data = tuned_results.set_index('Model')

# Create a bar chart
tuned_plot_data.plot(
    kind='bar',
    figsize=(10, 6)
)

# Add the chart title
plt.title("Tuned LightGBM vs Tuned XGBoost")

# Add the x axis label
plt.xlabel("Model")

# Add the y axis label
plt.ylabel("Score")

# Set the y axis range
plt.ylim(0, 1)

# Keep the model names horizontal
plt.xticks(rotation=0)

# Display the legend
plt.legend()

# Display the graph
plt.show()
```

```python
# Calculate tuned LightGBM metrics
tuned_lgbm_accuracy = accuracy_score(y_test, best_lgbm_pred)
tuned_lgbm_precision = precision_score(y_test, best_lgbm_pred)
tuned_lgbm_recall = recall_score(y_test, best_lgbm_pred)
tuned_lgbm_f1 = f1_score(y_test, best_lgbm_pred)

# Calculate tuned XGBoost metrics
tuned_xgb_accuracy = accuracy_score(y_test, best_xgb_pred)
tuned_xgb_precision = precision_score(y_test, best_xgb_pred)
tuned_xgb_recall = recall_score(y_test, best_xgb_pred)
tuned_xgb_f1 = f1_score(y_test, best_xgb_pred)

# Create a comparison DataFrame
tuned_comparison = pd.DataFrame({
    'Model': ['Tuned LightGBM', 'Tuned XGBoost'],
    'Accuracy': [
        tuned_lgbm_accuracy,
        tuned_xgb_accuracy
    ],
    'Precision': [
        tuned_lgbm_precision,
        tuned_xgb_precision
    ],
    'Recall': [
        tuned_lgbm_recall,
        tuned_xgb_recall
    ],
    'F1 Score': [
        tuned_lgbm_f1,
        tuned_xgb_f1
    ]
})

# Round the metrics
tuned_comparison = tuned_comparison.round(4)

# Display the tuned model comparison
print("Tuned Model Comparison")
print(tuned_comparison)
```

```python
# Set the Model column as the index
tuned_plot = tuned_comparison.set_index('Model')

# Create a bar chart
tuned_plot.plot(
    kind='bar',
    figsize=(10, 6)
)

# Add the title
plt.title('Tuned LightGBM vs Tuned XGBoost')

# Add the x axis label
plt.xlabel('Model')

# Add the y axis label
plt.ylabel('Performance Score')

# Set the y axis range
plt.ylim(0, 1)

# Keep model names horizontal
plt.xticks(rotation=0)

# Display the legend
plt.legend(title='Metrics')

# Display the graph
plt.show()
```

```python
# Compare the models for each evaluation metric
for metric in ['Accuracy', 'Precision', 'Recall', 'F1 Score']:

    # Find the model with the highest value for the current metric
    best_model = tuned_comparison.loc[
        tuned_comparison[metric].idxmax(),
        'Model'
    ]

    # Find the highest value for the current metric
    best_value = tuned_comparison[metric].max()

    # Display the result
    print(
        metric,
        ":",
        best_model,
        "with a score of",
        round(best_value, 4)
    )
```
